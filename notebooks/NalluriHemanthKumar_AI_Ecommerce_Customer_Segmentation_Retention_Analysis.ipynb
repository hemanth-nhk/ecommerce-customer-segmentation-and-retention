## 2. Raw Data Integrity

The raw Online Retail II file is treated as immutable. Its expected SHA-256 is frozen in the project audit.

The project uses:

- unique non-cancelled invoices as purchase events;
- `Quantity > 0`;
- `Price > 0`;
- non-null customer identifiers for customer-level analysis;
- exact duplicate removal before final feature extraction;
- cancellations retained separately as behavioral information;
- anonymous transactions excluded from customer-level modeling.


```python
def sha256_file(path: Path, chunk_size: int = 1024 * 1024) -> str:
    digest = hashlib.sha256()
    with path.open("rb") as f:
        while chunk := f.read(chunk_size):
            digest.update(chunk)
    return digest.hexdigest()

if RAW_DATA.exists():
    observed_sha = sha256_file(RAW_DATA)
    print("Raw file:", RAW_DATA)
    print("SHA-256:", observed_sha)
    print("Expected:", EXPECTED_RAW_SHA256)
    assert observed_sha == EXPECTED_RAW_SHA256, "Raw dataset hash mismatch."
    print("PASS: raw dataset integrity verified.")
else:
    print("Raw dataset not present in this submission environment; repository audit remains the authoritative integrity record.")
```
